### Loops vs. list comprehensions vs. NumPy

Same task, let's try to convert a bunch of measurements in deg C to deg F
three ways. Watch how the *style* changes even though two of these are doing
exactly the same thing under the hood.


```python
#preparing some data
# let's try to convert deg C to deg F
daily_temps = [12.5, 15.0, 9.8, 22.3, 18.1,23.5,27.3]  # a "week" of daily temps in deg Celsius

# Version 1: explicit loop

temps_f_loop = []
for t in daily_temps:
    temps_f_loop.append(t * 9/5 + 32)

print("loop:", temps_f_loop)
```

```python
# version 2: list comprehension
# Same result, more compact. IMPORTANT: this is still iterating under the hood —
# it is NOT faster than the loop above, just terser to write.
temps_f_comp = [t * 9/5 + 32 for t in daily_temps]

print("comprehension:", temps_f_comp)
```

```python
# Version 3: NumPy, vectorized
import numpy as np

#we convert our list to a numpy array
daily_temps_arr = np.array(daily_temps)

#now we do not have to interate through elements - we can apply calculation to the entire array at one go
temps_f_vec = daily_temps_arr * 9/5 + 32

print("vectorized:   ", temps_f_vec)

# Confirm all three approaches agree
print(temps_f_loop==temps_f_comp)
print(temps_f_loop==list(temps_f_vec))
```

Try running the conversion formula directly on the **plain list** (skipping the loop/comprehension).

### Timing the three approaches

At small scale the difference is invisible. Let's try it on a year of hourly data
(~8,760 values) and see what happens.


```python
# let's prepare some data
np.random.seed(0)
year_hourly_c = np.random.normal(loc=12, scale=6, size=8760)   # made up hourly temps, deg C
year_hourly_list = list(year_hourly_c)
year_hourly_list
```

```python
%%timeit -n1000
#checking time with loop - we will use %timeit decorator, but it needs to be a oneliner, so we create a function that executes the loop
result = []
for t in year_hourly_list:
    result.append(t * 9/5 + 32)
```

```python
print("List comprehension:")
%timeit -n1000 result=[t * 9/5 + 32 for t in year_hourly_list]
```

```python
print("NumPy vectorized:")
%timeit -n1000 year_hourly_c * 9/5 + 32
```

**Takeaway:** the loop and the comprehension come out roughly similar — comprehensions are a
*style* improvement, not a *performance* one. The vectorized NumPy version wins by a wide margin
because there's no explicit Python-level iteration at all — the whole array is processed at once.

### Missing data: `np.nan`

Missing values are a constant fact of life in observational climate data. NumPy represents
a missing float as `np.nan` ("Not a Number").


```python
temps_with_gap = np.array([12.5, 15.0, np.nan, 22.3, 18.1])

print(temps_with_gap)

#checking for nans
isnan=np.isnan(temps_with_gap) # boolean array flagging NaNs
print("nans",isnan)

#checking for valid data
isvalid=~np.isnan(temps_with_gap) # boolean array flagging NaNs, but then - inverting the boolean array
print("valid:",isvalid)

#calculating with nans

result=temps_with_gap*9/5+32
print(result) #still nan

#watch out for this
print(np.mean(temps_with_gap))           # careful — this propagates to NaN!
print(np.nanmean(temps_with_gap))        # this ignores NaNs
```

### Broadcasting

Broadcasting lets NumPy apply an operation between arrays of *different but compatible* shapes,
without writing an explicit loop. This is the concept that is a bit tricky.


```python
#preparing some data

# station_temps: shape (3 stations, 5 days)
station_temps = np.array([
    [12.5, 15.0, 9.8, 22.3, 18.1],
    [11.2, 14.8, 10.1, 21.9, 17.5],
    [13.0, 15.6, 9.2, 22.8, 18.6],
])

# a calibration offset per station: shape (3,)
station_offset = np.array([0.5, -0.2, 0.1])

# To apply a per-station offset correctly, we need the offset to align with the
# STATIONS axis, not the days axis. We do this by reshaping it to (3, 1):
station_offset_col = station_offset.reshape(-1, 1)
print(station_temps.shape)
print(station_offset_col.shape)

corrected = station_temps + station_offset_col
print(corrected)

```

```python
# Broadcasting lines up the LAST axis. Here that means the offset (3,) does NOT
# line up with the days axis (5,) — this will raise an error. Try it:
print(station_temps.shape)
print(station_offset.shape)
station_temps + station_offset

```

**Broadcasting rule of thumb:** NumPy compares shapes from the right. Dimensions are compatible
if they're equal, or if one of them is 1. `(3, 5)` and `(3, 1)` are compatible (the `1` gets
"stretched" to match `5`); `(3, 5)` and `(3,)` are NOT compatible, because `(3,)` is compared
against the trailing `5`, not the leading `3`.


### Array reshaping

```python
#preparing some data - daily temperature data for three weeks
daily_temps = np.array([12.5, 15.0, 9.8, 22.3, 18.1, 12.5, 17.1, 11.2, 14.8, 10.1, 21.9, 17.5, 17.9, 17.9, 13.0, 15.6, 9.2, 22.8, 18.6, 19.0, 21.1]
)

print(daily_temps.shape)

#now let's reshape
temps_by_week=daily_temps.reshape((3,7))
print(temps_by_week.shape)

#let's see if reshaping went well
print(daily_temps[0:7])
print(temps_by_week[0])

#this is when reshaping goes badly
temps_by_week_bad=daily_temps.reshape((7,3))
print(temps_by_week_bad.shape)

print(daily_temps[0:7])
print(temps_by_week_bad[:,0])
```

**Reshaping rule of thumb:** NumPy reshapes arrays from the right. So with the reshaping as above - it will take the first 7 values from the source array and put them into first row, etc. till it creates 3 rows.

### Array subsetting

```python
# let's work with our "by week" data
print(temps_by_week.shape)

#select first week
first_week=temps_by_week[0,:]
print(first_week.shape)

#select first weeks, but slightly different way
first_week=temps_by_week[0:1,:]
print(first_week.shape)

#select first two weeks
first_two_weeks=temps_by_week[0:2,:]
print(first_two_weeks.shape)
```

### Axis-based reductions

```python
#creating some data
# 4 stations - in rows, 6 days each - in columns
station_temps = np.array([
    [12.5, 15.0, 9.8, 22.3, 18.1,19.5],
    [11.2, 14.8, 10.1, 21.9, 17.5,19.6],
    [13.0, 15.6, 9.2, 22.8, 18.6,20.0],
    [13.2, 15.8, 9.8, 22.9, 18.3,20.5],
])

#over the entire array
print(station_temps.mean())

# shape: (stations, days)
print(station_temps.shape)

# we can calculate average along array axis:
print("Mean per station (average over days):", station_temps.mean(axis=1))
print("Mean per day (average over stations):", station_temps.mean(axis=0))
```

**Collapsing rule:** `axis=0` collapses the *first* (or *zeroth*) dimension (stations), leaving one value per the second dimension (day).
`axis=1` collapses the *second* dimension (days), leaving one value per the first dimesion (station).
